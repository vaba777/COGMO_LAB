In experiments, people often make more errors when they are fast. These relationships are very delicate and involve complex processes. Yet, we can make a fastman, slowman, middleman prediction to approximate the effects. Take the code from example 3 and adjust it such that it calculates both a time prediction AND an error prediction, as follows:\
A. The model starts with a basic error probability of 0.01 (or 1%)\
B. The total error likelihood is now a multiplication of factors:\
i. For each middleman step, the error probability is doubled (i.e., x 2)\
ii. For each slowman step, the error probability is halved (i.e., x 0.5)\
iii. For each fastman step, the error probability is tripled (i.e., x 3)\
C. Calculate for all possible combinations of fastman, middleman, slowman processes (3 step types with each 3 speed options; giving a total of 3 x 3 x 3 model outcomes) what the trial time and expected error probability is.\
D. Make a scatterplot that shows on the horizontal axis trial time and on the vertical axis error likelihood (either in a percentage, or as a fraction) This can be used to answer Brightspace question 1D.

```python
#define basic error probability as parameter set to 1%
basic_error_probability = 0.01
```

```python
#defining function where errors are included
FACTORS = {"fast": 3, "middle": 2, "slow": 0.5}

def _step(theta, speed="middle"):
    speed = speed if speed in FACTORS else "slow"  
    return theta[speed], FACTORS[speed]

def perceptualstep_error(speed="middle"):
    return _step(theta_p, speed)

def motorstep_error(speed="middle"):
    return _step(theta_m, speed)

def cognitivestep_error(speed="middle"):
    return _step(theta_c, speed)
```

```python
types = ["fast", "middle", "slow"]
types1 = ["fast", "middle", "slow"]
types2 = ["fast", "middle", "slow"]

results = {}
for p, m, c in product(types, types1, types2):
    t_p, f_p = perceptualstep_error(p)
    t_m, f_m = motorstep_error(m)
    t_c, f_c = cognitivestep_error(c)

    results[f"{p}_{m}_{c}"] = {
        "time": t_p + t_m + t_c,
        "error_factor": basic_error_probability * f_p * 2 * f_m * f_c * 2,
    }

        
```

```python
#rounding
#math.floor() for rounding down
#only if value>1
def conditional_rounding(value):
    if value >1:
        value = math.floor(value)
    return value

for v in results.values():
    for k, vv in v.items():
        if k == "error_factor":
            vv = conditional_rounding(vv)
            v.update({k:vv})

```
